```python

# ============================================================
# GROUNDNUT LEAF DISEASE CLASSIFICATION
# MobileNetV2 + Categorical Focal Loss + INT8 Quantization
# + Single-Core CPU Benchmarking
# ============================================================

# STEP 1: IMPORT LIBRARIES

import os
import time
import random
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt

from pathlib import Path
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, accuracy_score
from tensorflow import keras
from tensorflow.keras import layers
from tensorflow.keras.applications import MobileNetV2


# ============================================================
# STEP 2: PROJECT CONFIGURATION
# ============================================================

DATASET_DIR = r"C:\Users\vamsh\Downloads\dataset\A novel groundnut leaf dataset for detection and classification of groundnut leaf diseases"

IMAGE_SIZE = (224, 224)
BATCH_SIZE = 16

EPOCHS = 10
FINE_TUNE_EPOCHS = 5

LEARNING_RATE = 0.0001
SEED = 42

KERAS_MODEL_PATH = "groundnut_mobilenetv2.keras"
TFLITE_MODEL_PATH = "groundnut_mobilenetv2_int8.tflite"

random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)

print("TensorFlow Version:", tf.__version__)


# ============================================================
# STEP 3: SINGLE-CORE CPU CONFIGURATION
# ============================================================

try:
    tf.config.threading.set_intra_op_parallelism_threads(1)
    tf.config.threading.set_inter_op_parallelism_threads(1)
except RuntimeError:
    print("TensorFlow threading was already initialized.")


# ============================================================
# STEP 4: LOAD DATASET
# ============================================================

dataset_path = Path(DATASET_DIR)

if not dataset_path.exists():
    raise FileNotFoundError("Dataset directory not found!")

class_names = sorted([
    folder.name
    for folder in dataset_path.iterdir()
    if folder.is_dir()
])

NUM_CLASSES = len(class_names)

if NUM_CLASSES < 2:
    raise ValueError("At least two disease classes are required!")

print("\nNumber of Classes:", NUM_CLASSES)
print("Class Names:", class_names)

image_extensions = {".jpg", ".jpeg", ".png", ".bmp"}

image_paths = []
labels = []

for label, class_name in enumerate(class_names):

    class_folder = dataset_path / class_name

    files = [
        file for file in class_folder.rglob("*")
        if file.suffix.lower() in image_extensions
    ]

    print(class_name, ":", len(files), "images")

    if len(files) < 10:
        raise ValueError(
            f"Not enough images in {class_name}"
        )

    for file in files:
        image_paths.append(str(file))
        labels.append(label)

image_paths = np.array(image_paths)
labels = np.array(labels)

print("\nTotal Images:", len(image_paths))


# ============================================================
# STEP 5: DISPLAY SAMPLE IMAGES
# ============================================================

plt.figure(figsize=(12, 8))

for i in range(min(8, len(image_paths))):

    image = tf.keras.utils.load_img(
        image_paths[i],
        target_size=IMAGE_SIZE
    )

    plt.subplot(2, 4, i + 1)
    plt.imshow(image)
    plt.title(class_names[labels[i]])
    plt.axis("off")

plt.tight_layout()
plt.show()


# ============================================================
# STEP 6: TRAIN / VALIDATION / TEST SPLIT
# ============================================================

train_paths, temp_paths, train_labels, temp_labels = (
    train_test_split(
        image_paths,
        labels,
        test_size=0.40,
        random_state=SEED,
        stratify=labels
    )
)

val_paths, test_paths, val_labels, test_labels = (
    train_test_split(
        temp_paths,
        temp_labels,
        test_size=0.50,
        random_state=SEED,
        stratify=temp_labels
    )
)

print("\nDataset Split:")
print("Training Images:", len(train_paths))
print("Validation Images:", len(val_paths))
print("Testing Images:", len(test_paths))


# ============================================================
# STEP 7: IMAGE PREPROCESSING
# ============================================================

def preprocess_image(image_path, label):

    image = tf.io.read_file(image_path)

    image = tf.io.decode_image(
        image,
        channels=3,
        expand_animations=False
    )

    image.set_shape([None, None, 3])

    image = tf.image.resize(
        image,
        IMAGE_SIZE
    )

    image = tf.cast(image, tf.float32)

    image = tf.keras.applications.mobilenet_v2.preprocess_input(
        image
    )

    label = tf.one_hot(
        label,
        depth=NUM_CLASSES
    )

    return image, label


# ============================================================
# STEP 8: CREATE TENSORFLOW DATASETS
# ============================================================

def create_dataset(paths, labels, training=False):

    dataset = tf.data.Dataset.from_tensor_slices(
        (paths, labels)
    )

    if training:
        dataset = dataset.shuffle(
            len(paths),
            seed=SEED
        )

    dataset = dataset.map(
        preprocess_image,
        num_parallel_calls=tf.data.AUTOTUNE
    )

    if training:
        dataset = dataset.map(
            lambda x, y: (
                tf.image.random_flip_left_right(x),
                y
            ),
            num_parallel_calls=tf.data.AUTOTUNE
        )

    dataset = dataset.batch(BATCH_SIZE)

    dataset = dataset.prefetch(
        tf.data.AUTOTUNE
    )

    return dataset


train_ds = create_dataset(
    train_paths,
    train_labels,
    training=True
)

val_ds = create_dataset(
    val_paths,
    val_labels
)

test_ds = create_dataset(
    test_paths,
    test_labels
)

print("\nTensorFlow datasets created successfully!")


# ============================================================
# STEP 9: CATEGORICAL FOCAL LOSS
# ============================================================

class CategoricalFocalLoss(keras.losses.Loss):

    def __init__(self, gamma=2.0, alpha=None,
                 name="categorical_focal_loss"):

        super().__init__(name=name)

        self.gamma = float(gamma)
        self.alpha = alpha

    def call(self, y_true, y_pred):

        y_pred = tf.clip_by_value(
            y_pred,
            1e-7,
            1.0 - 1e-7
        )

        cross_entropy = -y_true * tf.math.log(y_pred)

        focal_weight = tf.pow(
            1.0 - y_pred,
            self.gamma
        )

        loss = focal_weight * cross_entropy

        if self.alpha is not None:
            loss = loss * tf.constant(
                self.alpha,
                dtype=tf.float32
            )

        return tf.reduce_mean(
            tf.reduce_sum(loss, axis=-1)
        )


print("\nCategorical Focal Loss created!")


# ============================================================
# STEP 10: BUILD MOBILENETV2 MODEL
# ============================================================

base_model = MobileNetV2(
    input_shape=(224, 224, 3),
    include_top=False,
    weights="imagenet"
)

base_model.trainable = False

inputs = keras.Input(
    shape=(224, 224, 3)
)

x = base_model(
    inputs,
    training=False
)

x = layers.GlobalAveragePooling2D()(x)

x = layers.Dropout(0.3)(x)

outputs = layers.Dense(
    NUM_CLASSES,
    activation="softmax"
)(x)

model = keras.Model(
    inputs=inputs,
    outputs=outputs
)

model.compile(
    optimizer=keras.optimizers.Adam(
        learning_rate=LEARNING_RATE
    ),
    loss=CategoricalFocalLoss(gamma=2.0),
    metrics=["accuracy"]
)

print("\nMobileNetV2 Model Summary:")

model.summary()


# ============================================================
# STEP 11: INITIAL MODEL TRAINING
# ============================================================

callbacks = [
    keras.callbacks.EarlyStopping(
        monitor="val_loss",
        patience=3,
        restore_best_weights=True
    )
]

print("\nStarting Initial Training...")

history = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=EPOCHS,
    callbacks=callbacks
)

print("\nInitial Training Completed!")


# ============================================================
# STEP 12: MOBILENETV2 FINE-TUNING
# ============================================================

print("\nStarting Fine-Tuning...")

base_model.trainable = True

for layer in base_model.layers[:-30]:
    layer.trainable = False

model.compile(
    optimizer=keras.optimizers.Adam(
        learning_rate=LEARNING_RATE / 10
    ),
    loss=CategoricalFocalLoss(gamma=2.0),
    metrics=["accuracy"]
)

fine_tune_history = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=FINE_TUNE_EPOCHS,
    callbacks=callbacks
)

print("\nFine-Tuning Completed!")


# ============================================================
# STEP 13: TRAINING ACCURACY GRAPH
# ============================================================

train_accuracy = (
    history.history["accuracy"] +
    fine_tune_history.history["accuracy"]
)

val_accuracy = (
    history.history["val_accuracy"] +
    fine_tune_history.history["val_accuracy"]
)

plt.figure(figsize=(10, 5))

plt.plot(
    train_accuracy,
    label="Training Accuracy"
)

plt.plot(
    val_accuracy,
    label="Validation Accuracy"
)

plt.xlabel("Epochs")
plt.ylabel("Accuracy")
plt.title("MobileNetV2 Training Accuracy")

plt.legend()
plt.grid(True)
plt.show()


# ============================================================
# STEP 14: TRAINING LOSS GRAPH
# ============================================================

train_loss = (
    history.history["loss"] +
    fine_tune_history.history["loss"]
)

val_loss = (
    history.history["val_loss"] +
    fine_tune_history.history["val_loss"]
)

plt.figure(figsize=(10, 5))

plt.plot(
    train_loss,
    label="Training Loss"
)

plt.plot(
    val_loss,
    label="Validation Loss"
)

plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.title("MobileNetV2 Training Loss")

plt.legend()
plt.grid(True)
plt.show()


# ============================================================
# STEP 15: MODEL EVALUATION
# ============================================================

print("\nEvaluating Model...")

test_loss, test_accuracy = model.evaluate(
    test_ds
)

predictions = model.predict(test_ds)

predicted_labels = np.argmax(
    predictions,
    axis=1
)

accuracy = accuracy_score(
    test_labels,
    predicted_labels
)

print("\nTest Accuracy:", round(accuracy * 100, 2), "%")

print("\nClassification Report:\n")

print(
    classification_report(
        test_labels,
        predicted_labels,
        labels=np.arange(NUM_CLASSES),
        target_names=class_names,
        zero_division=0
    )
)


# ============================================================
# STEP 16: SAVE TRAINED KERAS MODEL
# ============================================================

model.save(KERAS_MODEL_PATH)

print("\nKeras Model Saved Successfully!")

print("File:", KERAS_MODEL_PATH)


# ============================================================
# STEP 17: REPRESENTATIVE DATASET
# ============================================================

def representative_dataset_gen():

    sample_indices = np.arange(
        len(train_paths)
    )

    np.random.shuffle(sample_indices)

    sample_indices = sample_indices[
        :min(100, len(sample_indices))
    ]

    for index in sample_indices:

        image = tf.io.read_file(
            train_paths[index]
        )

        image = tf.io.decode_image(
            image,
            channels=3,
            expand_animations=False
        )

        image.set_shape([None, None, 3])

        image = tf.image.resize(
            image,
            IMAGE_SIZE
        )

        image = tf.cast(
            image,
            tf.float32
        )

        image = tf.keras.applications.mobilenet_v2.preprocess_input(
            image
        )

        image = tf.expand_dims(
            image,
            axis=0
        )

        yield [image.numpy()]


print("\nRepresentative Dataset Generator Created!")


# ============================================================
# STEP 18: INT8 TFLITE QUANTIZATION
# ============================================================

print("\nStarting INT8 Quantization...")

converter = tf.lite.TFLiteConverter.from_keras_model(
    model
)

converter.optimizations = [
    tf.lite.Optimize.DEFAULT
]

converter.representative_dataset = (
    representative_dataset_gen
)

converter.target_spec.supported_ops = [
    tf.lite.OpsSet.TFLITE_BUILTINS_INT8
]

converter.inference_input_type = tf.int8

converter.inference_output_type = tf.int8

tflite_model_quant = converter.convert()

with open(TFLITE_MODEL_PATH, "wb") as f:
    f.write(tflite_model_quant)

print("\nINT8 Conversion Completed!")

print("TFLite Model:", TFLITE_MODEL_PATH)


# ============================================================
# STEP 19: LOAD TFLITE MODEL
# ============================================================

interpreter = tf.lite.Interpreter(
    model_path=TFLITE_MODEL_PATH,
    num_threads=1
)

interpreter.allocate_tensors()

input_details = interpreter.get_input_details()

output_details = interpreter.get_output_details()

input_info = input_details[0]

output_info = output_details[0]

print("\nTFLite Model Loaded Successfully!")

print("Input Type:", input_info["dtype"])

print("Output Type:", output_info["dtype"])


# ============================================================
# STEP 20: SINGLE-CORE CPU BENCHMARKING
# ============================================================

print("\nStarting Single-Core CPU Benchmark...")

image = tf.io.read_file(
    test_paths[0]
)

image = tf.io.decode_image(
    image,
    channels=3,
    expand_animations=False
)

image.set_shape([None, None, 3])

image = tf.image.resize(
    image,
    IMAGE_SIZE
)

image = tf.cast(
    image,
    tf.float32
)

image = tf.keras.applications.mobilenet_v2.preprocess_input(
    image
)

image = tf.expand_dims(
    image,
    axis=0
).numpy()

scale, zero_point = input_info["quantization"]

if scale <= 0:
    raise ValueError("Invalid quantization scale!")

quantized_input = np.round(
    image / scale + zero_point
)

quantized_input = np.clip(
    quantized_input,
    -128,
    127
).astype(np.int8)


# Warm-up iterations

for _ in range(10):

    interpreter.set_tensor(
        input_info["index"],
        quantized_input
    )

    interpreter.invoke()


# Actual benchmark

iterations = 100

start_time = time.perf_counter()

for _ in range(iterations):

    interpreter.set_tensor(
        input_info["index"],
        quantized_input
    )

    interpreter.invoke()

end_time = time.perf_counter()

total_time = end_time - start_time

average_latency = (
    total_time / iterations
) * 1000

fps = 1000 / average_latency

print("\n========== CPU BENCHMARK RESULTS ==========")

print("CPU Threads: 1")

print("Iterations:", iterations)

print("Total Time:",
      round(total_time, 4), "seconds")

print("Average Inference Latency:",
      round(average_latency, 3), "ms")

print("Inference Throughput:",
      round(fps, 2), "images/second")

print("===========================================")


# ============================================================
# STEP 21: TEST INT8 MODEL PREDICTION
# ============================================================

interpreter.set_tensor(
    input_info["index"],
    quantized_input
)

interpreter.invoke()

output = interpreter.get_tensor(
    output_info["index"]
)

output_scale, output_zero_point = (
    output_info["quantization"]
)

if output_scale > 0:

    probabilities = (
        output.astype(np.float32) - output_zero_point
    ) * output_scale

else:

    probabilities = output.astype(np.float32)

predicted_class = np.argmax(
    probabilities[0]
)

confidence = float(
    probabilities[0][predicted_class]
) * 100

print("\n========== SAMPLE PREDICTION ==========")

print("Actual Class:", class_names[test_labels[0]])

print("Predicted Class:", class_names[predicted_class])

print("Prediction Confidence:",
      round(confidence, 2), "%")

print("=======================================")

plt.figure(figsize=(6, 5))

plt.imshow(
    tf.keras.utils.load_img(
        test_paths[0],
        target_size=IMAGE_SIZE
    )
)

plt.title(
    "Predicted: " + class_names[predicted_class]
)

plt.axis("off")

plt.show()


# ============================================================
# STEP 22: FINAL PROJECT SUMMARY
# ============================================================

keras_size = (
    os.path.getsize(KERAS_MODEL_PATH) /
    (1024 * 1024)
)

tflite_size = (
    os.path.getsize(TFLITE_MODEL_PATH) /
    (1024 * 1024)
)

size_reduction = (
    1 - tflite_size / keras_size
) * 100

print("\n========== FINAL PROJECT SUMMARY ==========")

print("Number of Classes:", NUM_CLASSES)

print("Class Names:", class_names)

print("Total Images:", len(image_paths))

print("Training Images:", len(train_paths))

print("Validation Images:", len(val_paths))

print("Testing Images:", len(test_paths))

print("Test Accuracy:", round(accuracy * 100, 2), "%")

print("Keras Model Size:", round(keras_size, 2), "MB")

print("INT8 Model Size:", round(tflite_size, 2), "MB")

print("Model Size Reduction:", round(size_reduction, 2), "%")

print("Average Inference Latency:",
      round(average_latency, 3), "ms")

print("Inference Throughput:",
      round(fps, 2), "images/second")

print("\nAll project stages completed!")

print("============================================")
```
